```python
def se_floor_uL(delta_m_g, rho_g_per_mL, n_reps, sigma_rep_g=None, use_quant_sd=True):
    """
    Return a sensible SE floor (epsilon) in µL for clipping.
    - delta_m_g: display resolution (g), e.g. 1e-4 g
    - rho_g_per_mL: liquid density
    - n_reps: repeats per level
    - sigma_rep_g: scale repeatability SD (g). If known, include it.
    - use_quant_sd: if True, use delta/√12 for quantization SD; else use delta directly (conservative).
    """
    # quantization SD (g)
    sigma_q = (delta_m_g/np.sqrt(12.0)) if use_quant_sd else delta_m_g
    # combine repeatability (if provided) with quantization
    if sigma_rep_g is not None:
        sigma_m = np.sqrt(sigma_rep_g**2 + sigma_q**2)
    else:
        sigma_m = sigma_q
    # two readings per dispense
    sigma_dm = np.sqrt(2.0) * sigma_m
    # convert to volume SD (µL), then SE of the mean
    sigma_V_uL = (sigma_dm / rho_g_per_mL) * 1000.0
    se_min_uL = sigma_V_uL / np.sqrt(n_reps)
    return se_min_uL
```

```python
import numpy as np

# --- inputs you provided ---
CF_current = 0.0600                 # mm/µL
BL_current = 0.3488880202852401     # mm
syringe_vol = 1000
n_points = 10   # how many data points
reps = 3    # how many times each point is measured

# set volumes (10%..100% capacity)
V_set = np.linspace(0.1*syringe_vol, syringe_vol, n_points)  # µL

# cumulative masses [g]; 31 points -> 30 dispenses
m_meas = np.array([
    0.0, 0.0966, 0.1932, 0.2900, 0.4865, 0.6831, 0.8798, 1.1753, 1.4710, 1.7667,
    2.1623, 2.5579, 2.9534, 3.4489, 3.9445, 4.4396, 5.0346, 5.6297, 6.2258, 6.9221,
    7.6183, 8.3144, 9.1107, 9.9072, 10.7036, 11.5999, 12.4964, 13.3928, 14.3891,
    15.3853, 16.3815
])
rho = 0.997926   # g/mL  (set your liquid density here)

# --- convert to per-dispense volumes ---
dm = np.diff(m_meas)                         # g
V_meas_all = (dm / rho) * 1000.0             # µL

# reshape to (n_points, reps): assumes your run order is 100,100,100, 200,200,200, ...
V_rep = V_meas_all.reshape(n_points, reps)
mean_uL = V_rep.mean(axis=1)
sd_uL   = V_rep.std(axis=1, ddof=1)
se_uL   = sd_uL / np.sqrt(reps)

# polyfit expects weights = 1/sigma_y; our y-points are the means -> use sigma = SE
eps_uL = se_floor_uL(delta_m_g=1e-4, rho_g_per_mL=rho, n_reps=reps)
w = 1.0 / np.clip(se_uL, eps_uL, None)

# --- weighted linear fit of measured mean vs set volume ---
# returns [slope, intercept]; i.e., V_meas ≈ a2 * V_set + b2
a2, b2 = np.polyfit(V_set, mean_uL, deg=1, w=w)

# --- update your inputs for next run ---
CF_new = CF_current / a2               # mm/µL
BL_new = BL_current - b2 * CF_new      # b2 is in µL -> convert to mm

print(f"a2={a2:.6f}, b2={b2:.6f} µL")
print(f"CF_new={CF_new:.6f} mm/µL, BL_new={BL_new:.6f} mm")
```

```python
def se_floor_uL(delta_m_g, rho_g_per_mL, n_reps, sigma_rep_g=None, use_quant_sd=True):
    """
    Return a sensible SE floor (epsilon) in µL for clipping.
    - delta_m_g: display resolution (g), e.g. 1e-4 g
    - rho_g_per_mL: liquid density
    - n_reps: repeats per level
    - sigma_rep_g: scale repeatability SD (g). If known, include it.
    - use_quant_sd: if True, use delta/√12 for quantization SD; else use delta directly (conservative).
    """
    # quantization SD (g)
    sigma_q = (delta_m_g/np.sqrt(12.0)) if use_quant_sd else delta_m_g
    # combine repeatability (if provided) with quantization
    if sigma_rep_g is not None:
        sigma_m = np.sqrt(sigma_rep_g**2 + sigma_q**2)
    else:
        sigma_m = sigma_q
    # two readings per dispense
    sigma_dm = np.sqrt(2.0) * sigma_m
    # convert to volume SD (µL), then SE of the mean
    sigma_V_uL = (sigma_dm / rho_g_per_mL) * 1000.0
    se_min_uL = sigma_V_uL / np.sqrt(n_reps)
    return se_min_uL
```
