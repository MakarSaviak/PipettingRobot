This is the file for the syringe/solvent database usage

```python
from sqlmodel import SQLModel, Session, create_engine
```

```python
from Syringe import Syringe
from Solvent import Solvent
from SyringeSolventLink import SyringeSolventLink
```

```python
engine = create_engine("sqlite:///liquid_handling.db", echo=False)
SQLModel.metadata.create_all(engine)
```

```python
with Session(engine) as s:
    obj1 = s.get(Syringe, 1)
    print(obj1)
    obj2 = Solvent(
        name="Acetonitrile",
        density_g_per_ml="0.780",
        notes="HPLC grade"
        )

    for obj in [obj1, obj2]:
        try:
            s.add(obj); s.commit(); s.refresh(obj)
        except Exception as e:
            s.rollback()
            print("Insert failed:", e)

    print(obj1)
    # make the link row (with metadata)
    link = SyringeSolventLink(syringe=obj1, solvent=obj2,
                          calibrated=False)
    s.add(link)
    s.commit()
```

```python
engine.dispose()
```
